# PART 2 — Broadcasting 
Broadcasting allows NumPy to perform operations on arrays with **different shapes automatically**.

Broadcasting Rules 
NumPy follows **3 rules**:
1. Compare shapes from **right to left**
2. Dimensions are compatible when:   - They are equal   - One of them is 1
3. Otherwise → **Broadcasting Error**

```python
A = np.array([[1,2,3],[4,5,6]])#2,3
print("A:",A)
print("-"*50)
B = np.array([10,20,30])
print("B:",B)
A + B
```

## 2. Shape CompatibilityTwo arrays are compatible when dimensions follow broadcasting rules.

```python
A = np.ones((2,3))#2,3
print(A)
print("-"*50)
B = np.ones((3,))#
print(B)
A + B
```

## 3. Scalar BroadcastingA **single value (scalar)** is automatically applied to every element.

```python
arr = np.array([1,2,3,4])
a1=arr + 5
print(a1)
```

```python
a2=arr * 10
print(a2)
```

## 4. Row‑wise Broadcasting A row vector operates across rows of a matrix.

```python
A = np.array([[1,2,3],[4,5,6]])#2,3
B = np.array([10,20,30])#(3,)
print(A)
print("-"*50)
print(B)
A + B
```

```python
A * B
```

## 5. Column‑wise Broadcasting A column vector operates across columns.

```python
A = np.array([[1,2,3],[4,5,6]])#2,3
B = np.array([[10],[20]])#2,1
print(A)
print(A.shape)
print("-"*50)
print(B)
print(B.shape)
print("-"*50)

A + B
```

```python
A * B
```

#(4,3)+(1,3)
#(5,4)+(4,)
#(3,2)+(6,3)


## 6. Broadcasting Errors & DebuggingBroadcasting fails when shapes are incompatible.

```python
A = np.array([[1,2,3],[4,5,6]])#2,3
B = np.array([10,20])#2 This will cause broadcasting error# A + B
A+B
```

### Fix using reshape

```python
B = np.array([10,20]).reshape(2,1)
print(B.shape)
A + B
```
