## loc[] and iloc[] in Pandas
Why loc and iloc are Important

In Data Science we often need to:

• select specific rows
• select specific columns
• filter data
• slice datasets

Pandas provides two powerful indexing methods:

1️⃣ .loc[] → Label-based indexing
2️⃣ .iloc[] → Position-based indexing



```python
#1. Import Pandas
import pandas as pd
#2. Create Sample Dataset
data = {
    "Name": ["Alice", "Bob", "Charlie", "David", "Emma"],
    "Age": [23, 25, 22, 24, 26],
    "Department": ["HR", "IT", "Finance", "IT", "Marketing"],
    "Salary": [40000, 50000, 45000, 52000, 48000]
}

df = pd.DataFrame(data)

print(df)


```

3. What is loc[] ?

.loc[] selects data using row labels and column labels.

Syntax
df.loc[row_label, column_label]

Example structure

df.loc[rows , columns]


```python
#4. Selecting Single Row using loc
df.loc[[2]]

```

```python

#5. Selecting Multiple Rows
df.loc[[1,3]]
```

```python
#6. Selecting Row and Column
df.loc[2, "Name"]




```

```python
#7. Selecting Multiple Columns
df.loc[2, ["Name", "Salary"]]




```

```python
#8. Selecting Multiple Rows and Columns
df.loc[[1,3], ["Name", "Department"]]


```

```python
#9. Row Slicing using loc

#Important rule:

#loc includes the end index
df.loc[1:3]


```

```python
#10. Conditional Selection using loc


#Example: Salary greater than 45000

df.loc[df["Salary"] > 45000]


#df["Salary"] > 45000
```

```python
#11. Selecting Specific Columns with Condition
df.loc[df["Salary"] > 45000, ["Name", "Salary"]]
```

```python
#12. Updating Values using loc
df.loc[2, "Salary"] = 47000

df
```

```python
#13. What is iloc[] ?


#.iloc[] selects data using integer positions (index numbers).

#Syntax
#df.iloc[row_position, column_position]
```

```python
#14. Selecting Single Row using iloc
df.iloc[2]

#Returns row at position 2.

```

```python
#15. Selecting Multiple Rows
df.iloc[[1,3]]
```

```python
#16. Selecting Row and Column
df.iloc[2,1]

#Explanation

#Row → position 2
#Column → position 1



```

```python
#17. Selecting Multiple Columns
df.iloc[2, [0,3]]


```

```python
#18. Selecting Multiple Rows and Columns
df.iloc[[1,3], [0,2]]


```

```python
#19. Row Slicing using iloc

#Important rule:

#iloc excludes the end index
df.iloc[1:4]



```

```python
#20. Column Slicing
df.iloc[:, 1:3]

```

```python

#21. Selecting Entire Column
df.iloc[:,1]


```

Difference Between loc and iloc

Feature	loc	        iloc

Type	Label based	Position based

Input	Index name	Integer position

Range	End included	End excluded

Usage	Filtering	Numerical indexing
