# Merging & Joining in Pandas

Pandas merge() 



merge() is used to combine two DataFrames based on common columns or indexes, similar to SQL joins.

"how"	Description

inner	Only rows with matching keys in both DataFrames

left	All rows from left DataFrame + matched rows from right

right	All rows from right DataFrame + matched rows from left

outer	All rows from both DataFrames (fills missing values with NaN)


 ## Sample Data

```python
import pandas as pd

df1 = pd.DataFrame({
    "EmpID": [1, 2, 3, 4],
    "Name": ["A", "B", "C", "D"],
    "DeptID": [101, 102, 103, 104]
})

df2 = pd.DataFrame({
    "DeptID": [101, 102, 105],
    "Department": ["HR", "IT", "Finance"]
})

df3 = pd.DataFrame({
    "EmpID": [1, 2, 5],
    "Salary": [50000, 60000, 70000]
})

print(df1)
print(df2)
print(df3)
```

## merge() Examples

```python
# Inner Join
pd.merge(df1, df2, on="DeptID", how="inner")
```

```python
# Left Join
pd.merge(df1, df2, on="DeptID", how="left")
```

```python
# Right merge
pd.merge(df1, df2, on="DeptID", how="right")
```

```python
# Outer merge
pd.merge(df1, df2, on="DeptID", how="outer")
```

```python
# Merge on EmpID
pd.merge(df1, df3, on="EmpID", how="inner")
```

```python
 

# Merge with suffix
df = pd.merge(df1, df3, on="EmpID", how="outer", suffixes=("_october", "_november"))
print("data frame october:",df1)
print()
# print("data frame 2:",df2)
# print()
print("data frame november:",df3)
print()
print(df)
```

## join() Examples

```python
# Index-based join
df1.set_index("EmpID").join(df3.set_index("EmpID"))
```

```python
# Left join using join
df1.join(df3.set_index("EmpID"), on="EmpID")
```

```python
# Other joins
df1.join(df3.set_index("EmpID"), on="EmpID", how="inner")
df1.join(df3.set_index("EmpID"), on="EmpID", how="outer")
```

## concat() Examples

```python
df4 = pd.DataFrame({
    "EmpID": [11, 12, 13, 14],
    "Name": ["G", "H", "K", "X"],
    "DeptID": [108, 109, 107, 108]
})
df4
```

```python
# Row-wise concat
pd.concat([df1, df4])#by default axis=0
```

```python
# Column-wise concat
pd.concat([df1, df4], axis=1)
```

```python
pd.concat([df1, df4],axis=0)
```

```python
# Ignore index
pd.concat([df1, df4], ignore_index=True)
```

```python
# Keys
pd.concat([df1, df4], keys=["First", "Second"])
```
