## 1️⃣ loc[] (Label-based indexing)

- Uses row and column labels
- Inclusive slicing
- Supports boolean filtering
- Used for meaningful indexing

Uses row labels (index) + column names

Syntax: df.loc[row_label, column_name]

End index is INCLUDED in slicing

Supports boolean filtering

Allows assignment (update values)

Can create new columns

Works well with real-world datasets

Supports multi-row and multi-column selection

Can use functions like isin(), between() inside loc

Preferred for data analysis tasks

Feature	loc[]	iloc[]

Type	Label-based	Position-based

Input	Row names, column names	Integer index

Slicing	Inclusive	Exclusive

Boolean filtering	✅ Yes	❌ No

Assignment	✅ Yes	❌ Limited

Readability	High	Medium

When to Use
✅ Use loc[] when:

Working with column names

Applying conditions

Updating/modifying data

Doing real-world analysis

✅ Use iloc[] when:

Working with positions

Doing quick slicing

Dataset has no meaningful labels
