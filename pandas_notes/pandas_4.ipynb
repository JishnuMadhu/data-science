# Pandas GroupBy

👉 Splits data into groups
Based on one or more columns (e.g., Department, Category)

👉 Applies operations on each group
Like sum, mean, count, max, min, etc.

👉 Combines results into a new structure
Returns a summarized DataFrame/Series

## Sample Data

```python
import pandas as pd

data = {
    "Department": ["HR", "IT", "IT", "HR", "Finance", "IT", "Finance"],
    "Employee": ["A", "B", "C", "D", "E", "F", "G"],
    "Salary": [40000, 60000, 65000, 45000, 70000, 62000, 72000],
    "Experience": [2, 5, 6, 3, 8, 4, 9],
    "City":["Bangalore","Chennai","Kochi","Kochi","Chennai","Kochi","Mumbai"]
}

df = pd.DataFrame(data)
df
```

## Basic groupby

```python
df.groupby("Department")["Salary"].mean()
```

```python
df.groupby("Department")["Salary"].sum()
```

```python
df.groupby("Department")['Salary'].max()
```

```python
df.groupby("Department")['Salary'].min()
```

```python
df.groupby("Department")['Employee'].count()
```

```python
df.groupby('City')['Employee'].count()
```

## Aggregations using agg()

```python
#average, maximum and minimum salary in each department
df.groupby("Department")["Salary"].agg(["mean", "max", "min"])
```

```python
#maximum and minimum experience in each department
df.groupby('Department')['Experience'].agg(['max','min'])
```

## Aggregations using agg() multiple columns

```python
df.groupby("Department").agg({
    "Salary": ["mean", "max"],
    "Experience": ["min", "max"]
})
```

## Named Aggregations

```python
df.groupby("Department").agg(
    Average_Salary=("Salary","mean"),
    Maximum_Salary=("Salary","max"),
Maximum_Experience=("Experience","max"),
Minimum_Experience=("Experience","min"),
)
```

```python
df.groupby("Department").agg(
    avg_salary=("Salary", "mean"),
    max_exp=("Experience", "max"),
    total_emp=("Employee", "count")
)
```

```python
df1.groupby(["Department", "Experience"])["Salary"].mean().reset_index()
```

## Transform
👉 Applies a function to each group

👉 Returns output with the SAME SIZE as original data

👉 Broadcasts group results back to each row

```python
df["Dept_Avg_Salary"] = df.groupby("Department")["Salary"].transform("mean")
df["Salary_Diff"] = df["Salary"] - df["Dept_Avg_Salary"]
df["Salary_Percent"] = df["Salary"] / df.groupby("Department")["Salary"].transform("sum")
df
```

```python
df["Dept_Avg_Salary"] = df.groupby("Department")["Salary"].transform("mean")
df
```
