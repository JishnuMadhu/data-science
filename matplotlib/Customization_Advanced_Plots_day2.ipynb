## What is a Subplot?

A subplot is a plot inside a grid of plots.

👉 Example: 2 rows × 2 columns = 4 plots in one figure

## Using plt.subplot()
plt.subplot(rows, cols, position)

```python
import matplotlib.pyplot as plt

plt.subplot(2, 2, 1)
plt.plot([1,2,3], [4,5,6],marker='o')
plt.title("Plot 1")

plt.subplot(2, 2, 2)
plt.plot([1,2,3], [6,5,4],marker='s')
plt.title("Plot 2")

plt.subplot(2, 2, 3)
plt.plot([1,2,3], [2,3,1],marker='^')
plt.title("Plot 3")

plt.subplot(2, 2, 4)
plt.plot([1,2,3], [3,1,2],marker='*')
plt.title("Plot 4")

plt.tight_layout()
plt.show()
```

## Using plt.subplots() 

Used to create all plots at once

fig, axs = plt.subplots(rows, cols)

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.arange(5)
print(x)

fig, axs = plt.subplots(1, 2)



axs[0].plot(x, x**2,marker='o',markerfacecolor='red')
axs[0].set_title("Square")
axs[0].set_xlabel("Numbers")
axs[0].set_ylabel("Square of numbers")

axs[1].plot(x, x**3,marker='s',markerfacecolor='green')
axs[1].set_title("Cube")
axs[1].set_xlabel("Numbers")
axs[1].set_ylabel("Cube of numbers")

plt.subplots_adjust(wspace=0.4,hspace=0.6)#adjusting space


plt.show()
```

```python
import matplotlib.pyplot as plt
plt.style.use('dark_background')
import numpy as np

# Days (X-axis)
days = np.arange(7)  # Days 0-6

# Sample stock prices
apple = [150, 152, 153, 151, 155, 157, 156]
google = [2800, 2820, 2810, 2830, 2840, 2850, 2860]
amazon = [3400, 3420, 3410, 3430, 3440, 3450, 3460]
microsoft = [299, 301, 300, 302, 303, 305, 304]

# Create 2x2 subplots
fig, axs = plt.subplots(2, 2, figsize=(10,8))

# Plot Apple
axs[0, 0].plot(days, apple, marker='o', color='red')
axs[0, 0].set_title("Apple Stock Price")
axs[0, 0].grid(True)

# Plot Google
axs[0, 1].plot(days, google, marker='s', color='blue')
axs[0, 1].set_title("Google Stock Price")
axs[0, 1].grid(True)

# Plot Amazon
axs[1, 0].plot(days, amazon, marker='^', color='green')
axs[1, 0].set_title("Amazon Stock Price")
axs[1, 0].grid(True)

# Plot Microsoft
axs[1, 1].plot(days, microsoft, marker='d', color='purple')
axs[1, 1].set_title("Microsoft Stock Price")
axs[1, 1].grid(True)


#Add overall X and Y labels
for ax in axs[1, :]:
    ax.set_xlabel("Day")
for ax in axs[:, 0]:
    ax.set_ylabel("Price ($)")


plt.tight_layout()
plt.show()
```

1️⃣ for ax in axs[1, :]
axs is a 2×2 array of subplots, i.e.,
axs[0,0]  axs[0,1]
axs[1,0]  axs[1,1]
axs[1, :] → all subplots in the second row
axs[1,0] → bottom-left subplot
axs[1,1] → bottom-right subplot
ax.set_xlabel("Day") → adds X-axis label ("Day") to bottom row subplots



Usually, only the bottom plots need X-axis labels for clarity.
Top plots don’t need X labels (they share X-axis visually).

2️⃣ for ax in axs[:, 0]
axs[:, 0] → all subplots in the first column
axs[0,0] → top-left subplot
axs[1,0] → bottom-left subplot
ax.set_ylabel("Price ($)") → adds Y-axis label ("Price ($)") to leftmost column subplots



Usually, only the leftmost plots need Y-axis labels.
Right column can omit labels to avoid clutter.

## Sharing Axes

```python
fig, axs = plt.subplots(2, 1,sharex=True)#sharing x axis, sharing y axis

axs[0].plot([1,2,3], [3,2,1])
axs[1].plot([1,2,3], [1,2,3])

plt.show()
```

```python
fig, axs = plt.subplots(1, 2,sharey=True)#sharing x axis, sharing y axis

axs[0].plot([1,2,3], [3,2,1])
axs[1].plot([1,2,3], [1,2,3])

plt.show()
```

## 4. Stacked Plots & Area Charts
- `stackplot()` for stacked data
- `fill_between()` to fill area under curve

```python
x = np.arange(5)
y1 = [1,2,3,4,5]
y2 = [2, 2, 2, 5, 2]
print(x)


plt.stackplot(x, y1, y2)
plt.title('Stacked Plot')
plt.show()

plt.fill_between(x, y1)
plt.title('Area Chart')
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

# X-axis: Days
days = np.arange(5)  # Day 0,1,2,3,4

# Traffic from different sources (numbers = visitors)
organic = [50, 60, 70, 80, 90]       # Organic search
social = [30, 40, 50, 60, 30]        # Social media
referral = [20, 25, 15, 20, 25]      # Referral websites

# Create stackplot
plt.stackplot(days, organic, social, referral, labels=['Organic', 'Social', 'Referral'])

# Add title and legend
plt.title("Website Traffic by Source")
plt.xlabel("Day")
plt.ylabel("Number of Visitors")
#plt.legend()
plt.legend(loc='upper right')

# Show plot
plt.show()
```

## 5. Box Plots & Violin Plots
- Boxplot shows quartiles and outliers
- Violin plot shows distribution shape

📊 Boxplot – Full Theory (in Points)

🔹 1. What is a Boxplot?

A boxplot (box-and-whisker plot) is a graphical summary of data distribution

It shows:

Center (median)

Spread (range & quartiles)

Outliers

Used for quick comparison between datasets
🔹 2. Five-Number Summary (Core Concept)

A boxplot is based on:

Minimum → smallest value

Q1 (First Quartile) → 25% of data below this

Median (Q2) → middle value (50%)

Q3 (Third Quartile) → 75% of data below this

Maximum → largest value

🔹 3. Quartiles Explained

Q1 → divides lowest 25%

Q2 (Median) → divides data into two halves

Q3 → divides top 25%

👉 So data is split into 4 equal parts

🔹 4. Structure of Boxplot

Box → from Q1 to Q3 (middle 50% data)

Line inside box → Median

Whiskers → extend to min and max (within limit)

Dots → Outliers

🔹 5. Interquartile Range (IQR)

IQR = Q3 - Q1

Represents spread of middle 50% data

👉 Important for detecting outliers

🔹 6. Outliers

Outliers = unusually high or low values

Calculated using:

Lower limit = Q1 - 1.5 × IQR  

Upper limit = Q3 + 1.5 × IQR

Values outside this range → outliers (shown as dots)

🔹 7. What Boxplot Tells You

Median → center of data

Box size → variability

Whiskers → range

Outliers → extreme values

Skewness → data distribution shape

🔹 8. Skewness (Shape of Data)

Symmetric → median in center

Right skewed → median closer to bottom (long upper whisker)

Left skewed → median closer to top (long lower whisker)

🔹 9. Advantages of Boxplot

Simple summary of data

Easy comparison of multiple datasets

Shows outliers clearly

Works well for large data

Example

Consider this dataset:

Data = [2, 4, 6, 8, 10, 12, 14, 16]

Step 1: Find Median (Q2)
Middle values → 8 and 10
Q2 = (8 + 10) / 2 = 9

Step 2: Find Q1

Lower half:

[2, 4, 6, 8]
Middle values → 4 and 6
Q1 = (4 + 6) / 2 = 5

Step 3: Find Q3

Upper half:

[10, 12, 14, 16]
Middle values → 12 and 14
Q3 = (12 + 14) / 2 = 13
📦 Final Values
Quartile	Value
Q1	5
Q2 (Median)	9
Q3	13

```python
import matplotlib.pyplot as plt

# Data
data = [2, 4, 6, 8, 10, 12, 14, 16]

# Create boxplot
plt.boxplot(data)

# Add title and labels
plt.title("Boxplot Example")
plt.ylabel("Values")

# Show plot
plt.show()
```

## Box Plot


```python
data=[10,12,14,15,18,20,22,25,30,100]
```

```python
plt.boxplot(data,vert=False)
plt.title("Before Removing Outliers")
plt.show()
```

```python
Q1=np.percentile(data,25)
Q3=np.percentile(data,75)
print(Q1)
print(Q3)
IQR=Q3-Q1
print(IQR)
```

```python
lower_limit=Q1-(1.5*IQR)
upper_limit=Q3+(1.5*IQR)
print("Lower limit:",lower_limit)
print("Upper limit:",upper_limit)
```

```python
outliers=[x for x in data if x<lower_limit or x>upper_limit]
print(outliers)
```

```python
clean_data=[x for x in data if lower_limit<= x <= upper_limit]
clean_data
```
