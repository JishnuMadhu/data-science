```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
plt.subplot(2,2,1)
plt.plot([1,2,3],[4,5,6],marker = 'o')
plt.title("Plot 1")

plt.subplot(2,2,2)
plt.plot([1,2,3],[6,5,4],marker = 's')
plt.title("Plot 2")

plt.subplot(2,2,3)
plt.plot([1,2,3],[2,3,1],marker = '^')
plt.title("Plot 3")

plt.subplot(2,2,4)
plt.plot([1,2,3],[3,1,2],marker = '*')
plt.title("Plot 4")

plt.tight_layout()
plt.show()
```

```python
x = [0,1,2,3]
y1 = [1,8,0,10]
y2 = [10,20,30,40]
y3 = y2[::-1]
y4 = y1[::-1]

plt.subplot(2,2,1)
plt.plot(x,y1,marker = 'o')
plt.grid()

plt.subplot(2,2,2)
plt.plot(x,y2,marker = 'o')
plt.grid()

plt.subplot(2,2,3)
plt.plot(x,y3,marker = 'o')
plt.grid()

plt.subplot(2,2,4)
plt.plot(x,y4,marker = 'o')
plt.grid()

plt.show()
```

```python
x = np.arange(5)
print(x)

fig, axs = plt.subplots(1,2)

axs[0].plot(x,x**2,marker = 'o',markerfacecolor = 'red')
axs[0].set_title('Square')
axs[0].set_xlabel('Numbers')
axs[0].set_ylabel('Square of numbers')
axs[0].grid()

axs[1].plot(x,x**3,marker = 'o',markerfacecolor = 'green')
axs[1].set_title('Cube')
axs[1].set_xlabel('Numbers')
axs[1].set_ylabel('Cube of Numbers')

plt.grid()
plt.tight_layout()
plt.show()
```
