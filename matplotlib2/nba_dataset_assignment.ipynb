```python
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv('nba.csv')
df.head()
```

```python
print(df.shape)       # rows and columns
print(df.columns)
```

```python
df.info() 
```

```python
print(df.shape)
```

```python
age_count = df['Age'].value_counts().sort_index()   # count players of each ag
plt.plot(age_count.index, age_count.values, linestyle='--', marker='o', color='green', label='Players')
plt.xlabel("Age")
plt.ylabel("Number of Players")
plt.title("Number of Players at Each Age")

plt.legend()
plt.show()
```

```python
plt.scatter(df['Age'], df['Weight'], s=15, color='orange', label='Players')
plt.xlabel("Age")
plt.ylabel("Weight (lbs)")
plt.title("Age vs Weight")

plt.legend()
plt.show()
```

```python
pos_count = df['Position'].value_counts()
pick_color = ['green', 'red', 'purple', 'orange', 'blue']
plt.bar(pos_count.index, pos_count.values, color=pick_color, edgecolor='black', label='Players')
plt.xlabel("Position")
plt.ylabel("Number of Players")
plt.title("Number of Players in Each Position")

plt.legend()
plt.show()
```

```python
plt.hist(df['Age'], bins=10, edgecolor='red', color='green', label='Players')
plt.xlabel("Age")
plt.ylabel("Frequency")
plt.title("Age Distribution of NBA Players")

plt.legend()
plt.show()
```

```python
position_count = df['Position'].value_counts()
hlt = [0.2,0,0,0,0]
plt.pie(position_count.values,labels=position_count.index,autopct='%1.1f%%',explode = hlt)
plt.title('Distribution of NBA Players by Position')
plt.show()
```
