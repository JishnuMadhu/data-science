```python
#1. Import Pandas
import pandas as pd
#2. Create Sample Dataset
data = {
"Name": ["Alice", "Bob", "Charlie", "David", "Emma"],
"Age": [23, 25, 22, 24, 26],
"Department": ["HR", "IT", "Finance", "IT", "Marketing"],
"Salary": [40000, 50000, 45000, 52000, 48000]
}

df = pd.DataFrame(data,index=[1,2,3,4,5])

print(df)
```

```python
#if there is multiple results, ti will not print in the cell,so we have to use the print function to do that
```

```python
print(df.loc[[2]])
print('*'*50)
print(df.iloc[[2]])
```

```python
df_new = pd.DataFrame(data,index=['A','B','C','D','E'])
df_new
```

```python
print(df_new.iloc[1,3])
```

```python
df_new.iloc[1:3,-1]
```
