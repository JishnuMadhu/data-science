```python
import pandas as pd

students = pd.DataFrame({
"student_id": [101, 102, 103, 104, 105, 106],
"student_name": ["Anu", "Rahul", "Meera", "Arun", "Diya", "Vivek"],
"course_id": ["C01", "C02", "C03", "C04", "C05", "C06"],
"marks": [85, 78, 92, 67, 88, 73]
})

courses = pd.DataFrame({
"course_id": ["C01", "C02", "C03", "C04", "C07", "C08"],
"course_name": [
"Python",
"Data Science",
"Machine Learning",
"Power BI",
"Generative AI",
"Cloud Computing"
],
"trainer": ["Anjali", "Rahul", "Meera", "Arun", "Neha", "Vishnu"]
})
```

```python
print(students)
print("*"*50)
print(courses)
```

```python
# Q1

#Create a report that shows only students who have a valid course assigned to them.
#Display the student name, course name, trainer, and marks.
```

```python
result = pd.merge(students,courses,on='course_id',how='inner')
print(result[['student_name','course_name','trainer','marks']])
```
